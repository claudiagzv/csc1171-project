## 9. Checking Available Activity Measurements

This code counts how many activity measurements are available for each organism and sorts them from highest to lowest.

```python
activity_columns = [
    "pf_ec50_uM",
    "tb_ic50_uM",
    "tc_ic50_uM",
    "tbrucei_ic50_uM",
    "ld_ic50_uM",
    "cp_ec50_uM",
    "tg_ec50_uM"
]

mmv[activity_columns].count().sort_values(ascending=False)
```

**Output:**

P. falciparum has the most activity measurements (1,687), followed by T. brucei (469) and T. cruzi (441). C. parvum has the fewest, with only 39.

This shows that P. falciparum has the most data available, making it a good starting point for machine learning. No changes were made to the dataset.

## 10. Visualising Available Activity Measurements

This code creates a bar chart to compare the number of activity measurements available for each organism.

```python
import matplotlib.pyplot as plt

activity_counts = mmv[activity_columns].count().sort_values(ascending=False)

activity_counts.plot(kind="bar")

plt.title("Number of Activity Measurements by Organism")
plt.xlabel("Activity Target")
plt.ylabel("Number of Compounds")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

**Output:**

The graph shows that P. falciparum has the most measurements (1,687), while C. parvum has the fewest (39).

This makes it easier to see the difference in data availability between organisms and shows why P. falciparum may be the best starting point for machine learning. No changes were made to the data.

## 17. Visualising P. falciparum EC50 on a Log Scale

This code removes missing EC50 values from a separate copy of the data and uses a log10 scale to make the distribution easier to see.

```python
import numpy as np

pf_data = mmv["pf_ec50_uM"].dropna()

log_pf_ec50 = np.log10(pf_data)

plt.hist(log_pf_ec50, bins=30)
plt.title("Distribution of P. falciparum EC50 (Log10 Scale)")
plt.xlabel("log10 EC50 (µM)")
plt.ylabel("Number of Compounds")
plt.grid(True, alpha=0.5)
plt.gca().set_axisbelow(True)
plt.show()
```

**Output:**

The graph shows a much clearer distribution, with most values grouped around 0 on the log10 scale. The extremely high EC50 values no longer stretch the graph as much.

This suggests that a log transformation could be useful when preparing the data for machine learning. The original dataset was not changed.
